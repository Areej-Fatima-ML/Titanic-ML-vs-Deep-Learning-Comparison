Task 2 : Data Processing (clean and prepare data)

1. Dropping Redundant Features

```python
# Dropping columns
cols_to_drop=['deck','embark_town','alive','who','class']
df_cleaned = df.drop(cols_to_drop, axis=1)
print(f"Columns remaining after dropping: { df_cleaned.columns.tolist()}")
```

Here dropping duplicate columns because derived from other columns.There is no need of redundent data .

2. Handling Missing values (Imputation)

```python
# Filling age with median
df_cleaned['age']=df_cleaned['age'].fillna(df_cleaned['age'].median())

# Filling embarked with mode
df_cleaned['embarked']=df_cleaned['embarked'].fillna(df_cleaned['age'].mode()[0])

# Final check after imputation
print(f"Missing values after imputation:  {df_cleaned.isnull().sum()}")
```

In this step imputed the age column with median because in statistical summary observed this column have outliers and mean (average) is very sensitive to  outliers.And embarked which is contains categorical text data so filled missing values with mode and complete dataset for model training.

3. Encoding categories

```python
# One-Hot Encoding
df_cleaned = pd.get_dummies(df_cleaned, columns=['sex', 'embarked'], drop_first=True,dtype=int)
# Converting other true false categories into binary (0/1)
df_cleaned=df_cleaned.astype(int)
# Result check
print("New columns after Encoding :")
print(df_cleaned.columns)
display(df_cleaned.head())
```

By using One-Hot encoding convert categorical text data into numbers because machine only understand numbers.And also change other True/False into binary 0/1 to ensure entire dataset into numeric format.

4. Train-Test Split (80/20)

```python
# Defining features(X) and Target(y)
X = df_cleaned.drop('survived', axis=1)
y = df_cleaned['survived']
# Splitting dataset into training and testing
from  sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,stratify=y,random_state=42)
# Checking shape of train and test
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
```

First define features and target we want to predict and then split dataset into training and testing set for so that model not get overfit(memorize data) and on test set evaluate performance to ensure model in actual learning patterns.used stratify so that target divide in equal ratio for training and testing.

5 . Feature Scaling


```python
# Scaling features
from sklearn.preprocessing import StandardScaler
# Initialzing StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
print("Scaling successfully completed")
```

Scaled features so that model give equal importance to all numbers not only prioritize  large numbers and we fit transform only training data to aviod  data leakage and to ensure use rules learn from  training set.

Task 3 : Training Machine Learning Models

1. Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.metrics import precision_score, recall_score, f1_score
# Training the model
lr_model=LogisticRegression(max_iter=1000)
lr_model.fit(X_train_scaled,y_train)
# Predictions
y_pred_lr=lr_model.predict(X_test_scaled)
# Accurcay scores
lr_train_acc = lr_model.score(X_train_scaled, y_train)
lr_test_acc = accuracy_score(y_test,y_pred_lr)

# Printing Accuracy score for both training  & testing
print(f" Logistic Regression Training Accuracy: {lr_train_acc:.4f}")
print(f" Logistic Regression Testing Accuracy: {lr_test_acc:.4f}")

# printing classification report
print(" Logistic Regression Classification Report:")
print(classification_report(y_test, y_pred_lr))

# Saving metrics for test set
lr_precision=precision_score(y_test,y_pred_lr)
lr_recall=recall_score(y_test,y_pred_lr)
lr_f1=f1_score(y_test,y_pred_lr)


# Confusion Matrix Heatmap
plt.figure(figsize=(5,4))
sns.heatmap(confusion_matrix(y_test,y_pred_lr), annot=True,fmt='d',cmap='Blues',
            xticklabels=['Predicted: Dead' , 'Predicted : Survived'],
            yticklabels=['Actual : Dead','Actual : Survived'])
plt.title('Confusion Matrix  : Logistic Regression')
plt.show()

# ROC Curve Calculation
from sklearn.metrics import roc_curve, auc
y_prob_lr=lr_model.predict_proba(X_test_scaled)[:,1]
fpr_lr, tpr_lr, _ = roc_curve(y_test, y_prob_lr)
roc_auc_lr = auc(fpr_lr, tpr_lr)
# Plotting
plt.figure(figsize=(5, 4))
plt.plot(fpr_lr, tpr_lr,label=f'ROC curve (area={roc_auc_lr:.2f})')
plt.plot([0, 1], [0, 1], 'k--')
plt.title('ROC Curve : Logistic Regression')
plt.legend()
plt.show()
```

Decision Boundary:

This model draw a linear line (straight) which seperate the two classes (Survived vs Dead) .It is called "Linear" because it is not a curve it is simply  mathematically divide data.

Task 4  Build and Train a Dense Neural Network

```python
# Importing libararies
import tensorflow as tf
# For Model structure
from tensorflow.keras.models import Sequential
# For layers and Regularization
from tensorflow.keras.layers import Dense,Dropout,Input
# Specific Optimizer
from tensorflow.keras.optimizers import Adam
# For control training
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
import matplotlib.pyplot as plt
print("Libraries imported successfully")


```

Model Architecture & Summary

```python
# Model structure (Architecture)
model = Sequential([

    # Input+1st Hidden layer (ReLU activation)  here scaled features (Age ,Sex,Pclass) entering into network+And first hidden layer
    # which use ReLU activation function and it's work is to find complex patterns.

    Input(shape=(X_train_scaled.shape[1],)),

    Dense(32, activation='relu'),

    # Reularization (Dropout rate between 0.2-0.4) this layer randomly off 30% neurons purpose it to prevent model from overfitting (memorizing data).

    Dropout(0.3),

    # 2nd Hidden layer (ReLU activation)  this layer learns more deep features patterns ReLU activation do the negative noise zero.
    Dense(16, activation='relu'),

    # Output Layer (Sigmoid for binary classification) this layer contains 1 neuron and sigmoid function and calculate probability (between 0 and 1)
    Dense(1, activation='sigmoid')])
#  Printing Model Summary
model.summary()
```

Layer 1 Dense  (Input and Hidden)  take inputs and learn pattern .And ReLU function present inside this layer.

Dropout preventing from overfitting (off some neurons).and learn patterns correctly.

Dense (Hidden) second layer further refine Learned patterns.

Dense(Output) use sigmoid function calculate probability for binary classification

Compiling and Training with Callbacks


```python
# Model Compilation
model.compile(optimizer=Adam(learning_rate=0.001),loss='binary_crossentropy',metrics=['accuracy'])
# EarlyStopping Callback
early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
# Model Training
history = model.fit(X_train_scaled, y_train, validation_data=(X_test_scaled, y_test), epochs=100, batch_size=32, callbacks=[early_stopping],verbose=1)
print("Model trained ")
```

Optimizer (Adam) in Backwardpropagation adjuts the weights smartly and fastly and keep learning rate not very large to
maintainlearning  balanace.
Loss Function (Binary Crossentropy):
It measures the error of model (difference between actual and prediction).

In Early Stopping callback monitor model validation loss if after 10 epochs model not get improve(minimize loss) training will automatically stop.After training Restore best weight move the model back to that where model performance was better.

Epochs 100 mean model seen data 100 times batch size (32) model at a time take 32 rows as inputs.
Validation data after every every epoch test on unseen data so that it's performnce can be check.







Training Progress Visualization

```python
# Plotting Accuracy and Loss Curves
plt.figure(figsize=(12,5))
# Accurcay Plot
plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label='Train Accuracy',color='teal')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy',color='orange')
plt.title('Training vs Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

# Loss Plot
plt.subplot(1,2,2)
plt.plot(history.history['loss'], label='Train Loss',color='teal')
plt.plot(history.history['val_loss'], label='Val Loss',color='orange')
plt.title('Training vs Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

Observed in Accuracy curve analysis both training and validation moving upward indicating model learning correctlt.Dropout and EarlyStopping ensures both lines remain close to each other .

 Observed In loss curve analysis both training and testing at same time moving downward  indicating Backpropagation and Adam Optimizer adjusting in right direction.
